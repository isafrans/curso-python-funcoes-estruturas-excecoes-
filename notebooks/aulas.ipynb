### Importando bibliotecas

```python
# Importação simples
import random

# Importação com alias
import numpy as np
import pandas as pd

# Importação de função específica
from random import choice

print(random.randint(1, 10))
print(choice(['maçã', 'banana', 'uva']))
```

### Utilizando pacotes — exemplos práticos

```python
# Exemplo com numpy
notas = np.array([7.5, 8.0, 6.5, 9.0, 7.0])
print(f'Média: {np.mean(notas):.2f}')
print(f'Maior nota: {np.max(notas)}')
print(f'Menor nota: {np.min(notas)}')
```

## 🔧 Módulo 2 — Funções

### Funções built-in

```python
notas = [7.5, 8.0, 6.5, 9.0, 7.0]

# Usando funções embutidas
media = sum(notas) / len(notas)
print(f'Média calculada com sum/len: {media:.2f}')
print(f'Máximo: {max(notas)} | Mínimo: {min(notas)}')
```

### Criando funções personalizadas

```python
def calcular_media(lista_notas):
    """Calcula a média de uma lista de notas.
    
    Args:
        lista_notas (list): Lista com as notas numéricas.
    
    Returns:
        float: Média das notas.
    """
    return sum(lista_notas) / len(lista_notas)


def calcular_estatisticas(lista_notas):
    """Retorna média, maior e menor nota."""
    media = calcular_media(lista_notas)
    maior = max(lista_notas)
    menor = min(lista_notas)
    return media, maior, menor


media, maior, menor = calcular_estatisticas(notas)
print(f'Média: {media:.2f} | Maior: {maior} | Menor: {menor}')
```

### Funções lambda e map()

```python
# Lambda para verificar aprovação
aprovado = lambda nota: 'Aprovado' if nota >= 7 else 'Reprovado'

# Aplicando com map()
situacoes = list(map(aprovado, notas))
print(situacoes)
```

## 🗂️ Módulo 3 — Estruturas de Dados

### Lista de listas e tuplas

```python
# Lista de listas
alunos = [
    ['Ana', 8.5, 9.0],
    ['Bruno', 6.0, 7.5],
    ['Carla', 9.5, 8.0]
]

# Indexação aninhada
print(f'Primeira nota de Bruno: {alunos[1][1]}')

# Lista de tuplas
alunos_tuplas = [('Ana', 8.5), ('Bruno', 6.0), ('Carla', 9.5)]
for nome, nota in alunos_tuplas:
    print(f'{nome}: {nota}')
```

### List comprehension

```python
nomes = ['Ana', 'Bruno', 'Carla']
notas_turma = [8.5, 6.0, 9.5]

# List comprehension com if-else
situacao = ['Aprovado' if n >= 7 else 'Reprovado' for n in notas_turma]
print(situacao)

# Com zip()
resultado = [(nome, sit) for nome, sit in zip(nomes, situacao)]
print(resultado)
```

### Dict comprehension

```python
# Dict comprehension — aluno: situação
boletim = {nome: sit for nome, sit in zip(nomes, situacao)}
print(boletim)
```

## ⚠️ Módulo 4 — Lidando com Exceções

### try / except

```python
dados = {'Ana': 8.5, 'Bruno': 6.0}

aluno = 'Carlos'
try:
    nota = dados[aluno]
    print(f'Nota de {aluno}: {nota}')
except KeyError:
    print(f'Aluno "{aluno}" não encontrado no dicionário.')
```

### finally e raise

```python
def validar_nota(nota):
    """Valida se a nota está no intervalo de 0 a 10."""
    if not (0 <= nota <= 10):
        raise ValueError(f'Nota inválida: {nota}. Deve estar entre 0 e 10.')
    return nota


try:
    validar_nota(11)
except ValueError as e:
    print(f'Erro: {e}')
finally:
    print('Validação concluída.')
```
